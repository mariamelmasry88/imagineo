## 5. start the website
this prints a public link. open it in your browser. the models load in the background, a banner on the page tells you when it is ready (a few minutes).

```python
import subprocess, sys, time
from pyngrok import ngrok

try:
    proc.terminate()
except NameError:
    pass
ngrok.kill()
ngrok.set_auth_token(UserSecretsClient().get_secret("ngrok"))
log = open("server.log", "w")
proc = subprocess.Popen([sys.executable, "-m", "uvicorn", "server:app", "--host", "127.0.0.1", "--port", "8000"], stdout=log, stderr=subprocess.STDOUT)
time.sleep(6)
print(ngrok.connect(8000).public_url)
```
